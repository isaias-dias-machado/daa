## Logistic Regression

Grid Search is used to optimize the regularization strength (`C`) and class weighting (`class_weight`).

```python
import pandas as pd

from daa.config import OUTPUTS_DIR
from daa.data import load_train
from daa.experiment import run_experiment
from daa.splits import get_folds
```


### Evaluate Logistic Regression using the selected features and fit a final model on the complete training dataset.



```python
X, y = load_train()
folds = get_folds(y)
print(f"real data: {X.shape[0]:,} rows x {X.shape[1]:,} columns")
print(f"{len(folds)} folds, {y.nunique()} classes")
```

```python
LogisticRegression = run_experiment(
    "logreg",
    name="Logistic Regression",
    X=X,
    y=y,
    folds=folds,
    feature_set="selected",
    fit_full=True,
)
pd.Series(LogisticRegression.metrics, name=LogisticRegression.name)
```
