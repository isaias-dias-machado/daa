### Optimize Random Forest, Logistic Regression and XGBoost hyperparameters using GridSearchCV.
### Uses the selected features and the full frozen cross-validation folds.

```python
import json

import pandas as pd
from sklearn.calibration import CalibratedClassifierCV
from sklearn.linear_model import LogisticRegression

from daa.config import OUTPUTS_DIR, RUNS_DIR
from daa.data import load_test, load_train
from daa.evaluation import load_run, validate_submission, write_submission_from_result
from daa.experiment import run_experiment
from daa.models.registry import MODEL_NAMES
from daa.models.spaces import SEARCH_SPACES
from daa.splits import get_folds
```

```python
X, y = load_train()
folds = get_folds(y)
print(f"real data: {X.shape[0]:,} rows x {X.shape[1]:,} columns")
print(f"{len(folds)} folds, {y.nunique()} classes")
print(f"models: {', '.join(MODEL_NAMES)}")
print(f"with search spaces: {', '.join(SEARCH_SPACES)}")
```

```python
rf_result = run_experiment(
    "rf",
    name="07_rf_selected_grid",
    X=X,
    y=y,
    folds=folds,
    feature_set="selected",
    search="grid",
)

print("Random Forest - GridSearchCV")
print(f"Macro F1: {rf_result.metrics['macro_f1_mean']:.4f}")
print(f"QWK:      {rf_result.metrics['qwk_mean']:.4f}")
```

```python
logreg_result = run_experiment(
    "logreg",
    name="07_logreg_selected_grid",
    X=X,
    y=y,
    folds=folds,
    feature_set="selected",
    search="grid",
)

print("Logistic Regression - GridSearchCV")
print(f"Macro F1: {logreg_result.metrics['macro_f1_mean']:.4f}")
print(f"QWK:      {logreg_result.metrics['qwk_mean']:.4f}")
```
