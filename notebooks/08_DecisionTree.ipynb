## Decision Tree

The Decision Tree model is evaluated using the selected features.

A full model is also fitted using the complete training dataset.

```python
import pandas as pd

from daa.config import OUTPUTS_DIR
from daa.data import load_train
from daa.experiment import run_experiment
from daa.splits import get_folds
```

## Data Loading

Load the training dataset and retrieve the predefined cross-validation folds.

```python
X, y = load_train()
folds = get_folds(y)
print(f"real data: {X.shape[0]:,} rows x {X.shape[1]:,} columns")
print(f"{len(folds)} folds, {y.nunique()} classes")
```

## Decision Tree

Evaluate the Decision Tree using the selected features and fit a final model on the complete training dataset.

```python
DecisionTree = run_experiment(
    "decision_tree",
    name="Decision Tree",
    X=X,
    y=y,
    folds=folds,
    feature_set="selected",
    fit_full=True,
)
pd.Series(DecisionTree.metrics, name=DecisionTree.name)
```

## Results

Retrieve the Decision Tree experiment results from the leaderboard, including the number of features and cross-validation performance.

```python
leaderboard = pd.read_csv(OUTPUTS_DIR / "leaderboard.csv")
score_columns = ["run_id", "name", "n_features", "macro_f1_mean", "macro_f1_std"]
leaderboard[leaderboard["name"].str.contains("decision tree", case=False)][
    score_columns
]
```
